# 7. Group contrast vs individual deviation -- Pancreatic Cancer (Moore et al.)

- Two sample arms, each paired with its own DESeq2 runs and shown with the same figures in the same order.

| | arm | samples (neoplasm / HC) | DESeq2 runs |
|---|---|---|---|
| **Part A** | Moore `Batch_1` + `Batch_2` + `Batch_3` | 72 / 69 | `pancreatic_deg_pooled`: our own DESeq2 (+ RUVg k=1-3) on the pooled cohort, not a reproduction of Moore et al. |
| **Part B** | Moore `Batch_1` only | 39 / 48 | `pancreatic_deg`: the CEDAR cohort Moore et al. fitted on, same designs |

- Each part: Signal Number Heatmap -> UpSet -> then, for the pancreas, liver and Open Targets panels in turn: cutoff sweep -> marker recovery -> heatmap -> Venn.
- NM-side samples are exactly the samples each DESeq2 run was fitted on (read back from its `ruvg_W/*.csv`), so the two sides cannot drift.
- Hit rules:
  - Tissue marker genes (pancreas, liver panels; `hpa_panels.tct_panel_table`): any detection, per-sample BH q<0.05 with Z>0, and in a patient >= MIN_HIT_READS (10) reads (HC hits carry no read condition).
  - Everything else (genome-wide unions, Open Targets genes, n_sig): two-sided BH q<0.05 and 10% detection.
- Part C holds the supporting tables and the Moore comparison.

```python
def load_source(src):
    if src == 'insample':
        ZD = config.ZSCORES_DIR
        dm = pd.read_csv(ZD / 'sample_meta.csv')
        hm = pd.read_csv(ZD / 'hc_meta.csv')
        dk = dm['ood_keep'].values.astype(bool)
        return dict(genes=np.array(pickle.load(open(ZD / 'gene_names.pkl', 'rb'))),
                    Z_dis=np.load(ZD / 'Z_disease_shash.npy')[dk],
                    dis=dm.loc[dk, 'sample'].values, dis_batch=dm.loc[dk, 'batch'].values,
                    Z_hc=np.load(ZD / 'Z_hc_shash.npy'), hc=hm['sample'].values,
                    hc_batch=hm['batch'].values)
    Zs, nm, bb, gl = [], [], [], None
    for b in MOORE_BATCHES:
        LD = config.LOBO_DIR / b.replace(' ', '_')
        info = json.load(open(LD / 'meta.json'))
        k = np.load(LD / 'ood_mask.npy').astype(bool)
        g = np.array(pickle.load(open(LD / 'gene_names.pkl', 'rb')))
        assert gl is None or np.array_equal(gl, g)
        gl = g
        Zs.append(np.load(LD / 'Z_test_shash.npy')[k])
        nm.append(np.array(info['test_names'])[k])
        bb.append(np.repeat(b, int(k.sum())))
    Z, nm, bb = np.vstack(Zs), np.concatenate(nm), np.concatenate(bb)
    ih = np.isin(nm, pd.read_csv(config.ZSCORES_DIR / 'hc_meta.csv')['sample'].values)
    return dict(genes=gl, Z_dis=Z[~ih], dis=nm[~ih], dis_batch=bb[~ih],
                Z_hc=Z[ih], hc=nm[ih], hc_batch=bb[ih])


src = load_source(SIG_SOURCE)
genes = src['genes']
gpos = {g: i for i, g in enumerate(genes)}

with h5py.File(config.H5AD_PATH, 'r') as h:
    dec = lambda a: np.array([x.decode('utf-8', 'replace') if isinstance(x, bytes) else str(x) for x in a])
    g = h['obs/Stage/Condition']
    stage_all = pd.Series(pd.Categorical.from_codes(g['codes'][()], dec(g['categories'][()])).astype(str),
                          index=dec(h['obs/_index'][()]))

dis_stage = pd.Series(src['dis']).map(stage_all).values
dis_idx, hc_idx = pd.Index(src['dis']), pd.Index(src['hc'])

# all arm: every Moore Batch_1-3 pancreatic sample the engine scored
a_moore = np.isin(src['dis_batch'], MOORE_BATCHES)
a_neo, a_panc = a_moore & np.isin(dis_stage, NEOPLASTIC), a_moore & (dis_stage == 'Pancreatitis')
a_hc = np.isin(src['hc_batch'], MOORE_BATCHES)
Z_A, names_A, stage_A = src['Z_dis'][a_neo], src['dis'][a_neo], dis_stage[a_neo]
Z_hc_A0, names_hc_A0 = src['Z_hc'][a_hc], src['hc'][a_hc]


def deg_case_samples(case):
    """The exact case+HC sample list the matched DESeq2 run was fitted on."""
    s = pd.read_csv(DEG_DIR / 'ruvg_W' / f'moore_b1__{case}.csv')['sample'].values
    st = pd.Series(s).map(stage_all).values
    return s, st


# matched arm: taken from the DESeq2 run rather than re-derived, so the two sides cannot drift
mt, mt_stage = deg_case_samples('PancreaticNeoplasm')
is_case = np.isin(mt_stage, NEOPLASTIC)
pat_m, hcm = dis_idx.get_indexer(mt[is_case]), hc_idx.get_indexer(mt[~is_case])
assert (pat_m >= 0).all() and (hcm >= 0).all(), 'DESeq2 sample missing from the Z matrices'
Z_neo0, names_neo0, stage_neo0 = src['Z_dis'][pat_m], src['dis'][pat_m], mt_stage[is_case]
Z_hc_m0, names_hc_m0 = src['Z_hc'][hcm], src['hc'][hcm]

print(f'genes={len(genes)}')
print(f'matched arm ({MATCHED_BATCH}, paired with {DEG_DIR.name}): '
      f'neoplasm={len(names_neo0)}  HC={len(names_hc_m0)}')
print(pd.Series(stage_neo0).value_counts().reindex(NEOPLASTIC).to_string())
print(f'all arm (Moore Batch_1-3): neoplasm={len(names_A)}  pancreatitis={int(a_panc.sum())}  '
      f'HC={len(names_hc_A0)}')
print(pd.Series(stage_A).value_counts().reindex(NEOPLASTIC).to_string())
```

```python
MIN_HIT_READS = 10
_read_ok = {}

def read_ok(names):
    miss = [s for s in names if s not in _read_ok]
    if miss:
        with h5py.File(config.H5AD_PATH, 'r') as h:
            obs = pd.Index(dec(h['obs/_index'][()]))
            var = dec(h['var/_index'][()])
            col = pd.Index(var).get_indexer(genes)
            assert (col >= 0).all()
            indptr = h['X/indptr'][()]
            for s in miss:
                i = obs.get_loc(s)
                row = np.zeros(len(var))
                row[h['X/indices'][indptr[i]:indptr[i + 1]]] = h['X/data'][indptr[i]:indptr[i + 1]]
                _read_ok[s] = row[col] >= MIN_HIT_READS
    return np.vstack([_read_ok[s] for s in names])


def bh_sig_matrix(Z, q=FDR_Q):
    return np.vstack([bh_fdr_reject(2 * norm.sf(np.abs(Z[i])), q=q) for i in range(Z.shape[0])])

def marker_sig_matrix(Z, q=FDR_Q, names=None):
    S = bh_sig_matrix(Z, q) & (Z > 0)
    return S if names is None else S & read_ok(names)

Z_neo, names_neo, stage_neo = Z_neo0, names_neo0, stage_neo0
Z_hc_qc, names_hc = Z_hc_m0, names_hc_m0
SIG, SIG_HC = bh_sig_matrix(Z_neo), bh_sig_matrix(Z_hc_qc)
SIG_M, SIG_HC_M = marker_sig_matrix(Z_neo, names=names_neo), marker_sig_matrix(Z_hc_qc)

Z_hc_A, names_hc_A = Z_hc_A0, names_hc_A0
SIG_A, SIG_HC_A = bh_sig_matrix(Z_A), bh_sig_matrix(Z_hc_A)
SIG_A_M, SIG_HC_A_M = marker_sig_matrix(Z_A, names=names_A), marker_sig_matrix(Z_hc_A)
print(f'matched arm: patients {len(names_neo)}, HC {len(names_hc)} | '
      f'all arm: patients {len(names_A)}, HC {len(names_hc_A)}')

HC_NSIG_CACHE = OUT / 'hc_pool_nsig.csv'
if HC_NSIG_CACHE.exists():
    hc_pool_nsig = pd.read_csv(HC_NSIG_CACHE)
else:
    hc_pool_nsig = pd.read_csv(config.ZSCORES_DIR / 'hc_meta.csv').assign(
        n_sig=bh_sig_matrix(np.load(config.ZSCORES_DIR / 'Z_hc_shash.npy')).sum(axis=1))
    hc_pool_nsig.to_csv(HC_NSIG_CACHE, index=False)

HC_NSIG_CUT = float(np.percentile(hc_pool_nsig['n_sig'], HC_NULL_PCT))
nsig_of = hc_pool_nsig.set_index('sample')['n_sig']
null_keep = pd.Series(names_hc).map(nsig_of).values <= HC_NSIG_CUT
Z_hc_null, names_hc_null = Z_hc_qc[null_keep], names_hc[null_keep]
SIG_HC_NULL = SIG_HC[null_keep]   # BH is per sample, so this is a row subset, not a refit
SIG_HC_NULL_M = SIG_HC_M[null_keep]
print(f'HC null pool: {null_keep.sum()}/{len(null_keep)} of the Batch_1 controls '
      f'(p{HC_NULL_PCT} of all {len(hc_pool_nsig)} HC = n_sig > {HC_NSIG_CUT:.0f})')
print('  disqualified:', {s: int(nsig_of[s]) for s in names_hc[~null_keep]})

per_sample = pd.DataFrame(dict(sample=names_A, stage=stage_A, n_sig=SIG_A.sum(axis=1),
                               in_matched=np.isin(names_A, names_neo)))
per_sample.to_csv(OUT / 'per_sample_nsig.csv', index=False)

if SIG_SOURCE == 'insample':
    ref = pd.read_csv(config.ZSCORES_DIR / 'sample_bh_sig.csv').set_index('sample')['n_sig']
    chk = per_sample.assign(pipeline=per_sample['sample'].map(ref))
    print(f'BH vs pipeline sample_bh_sig.csv: exact={int((chk.n_sig == chk.pipeline).sum())}/{len(chk)}')

print(per_sample.groupby('stage')['n_sig'].agg(n='size', median='median', mean='mean')
      .reindex(NEOPLASTIC).round(1).to_string())
print(f'matched-arm patients: n={len(names_neo)}, median n_sig={np.median(SIG.sum(axis=1)):.0f}')
print(f'matched-arm HC (n={len(names_hc)}): median n_sig = {np.median(SIG_HC.sum(axis=1)):.0f}, '
      f'mean = {SIG_HC.sum(axis=1).mean():.1f}')
print(f'all-arm HC (n={len(names_hc_A)}): median n_sig = {np.median(SIG_HC_A.sum(axis=1)):.0f}')
```

```python
def raw_counts(samples, ens_list):
    with h5py.File(config.H5AD_PATH, 'r') as h:
        obs_index = list(dec(h['obs/_index'][()]))
        vpos = {e: i for i, e in enumerate(dec(h['var/_index'][()]))}
        cols = np.array([vpos[e] for e in ens_list])
        indptr = h['X/indptr'][()]
        out = np.zeros((len(samples), len(cols)))
        lib = np.zeros(len(samples))
        for r, s in enumerate(samples):
            i = obs_index.index(s)
            sl = slice(indptr[i], indptr[i + 1])
            idx, dat = h['X/indices'][sl], h['X/data'][sl]
            lib[r] = dat.sum()
            pos = pd.Series(np.arange(len(idx)), index=idx).reindex(cols).values
            ok = ~pd.isna(pos)
            out[r, ok] = dat[pos[ok].astype(int)]
    return out, lib


ens_list = marker_tbl['ens'].tolist()
cnt, libsize = raw_counts(np.concatenate([names_neo, names_hc]), ens_list)
is_pat = np.arange(len(cnt)) < len(names_neo)
cpm = cnt / libsize[:, None] * 1e6

rows = []
for s in MARKER_SETS:
    c = marker_tbl['set'].values == s
    tot = cnt[:, c].sum(axis=1)
    cs = cpm[:, c].sum(axis=1)
    a, b = int((tot[is_pat] > 0).sum()), int((tot[~is_pat] > 0).sum())
    rows.append(dict(set=s, n_ens=int(c.sum()),
                     pat_detected=f'{a}/{is_pat.sum()}', hc_detected=f'{b}/{(~is_pat).sum()}',
                     pat_reads=int(tot[is_pat].sum()), hc_reads=int(tot[~is_pat].sum()),
                     pat_cpm=cs[is_pat].mean(), hc_cpm=cs[~is_pat].mean()))
detect = pd.DataFrame(rows)
detect.to_csv(OUT / 'marker_raw_detectability.csv', index=False)
print('is the transcript measurable at all? (raw counts, same samples)')
print(detect.round(4).to_string(index=False))

per_gene = pd.DataFrame(dict(
    set=marker_tbl['set'], symbol=marker_tbl['symbol'],
    pat_nz=(cnt[is_pat] > 0).sum(axis=0), hc_nz=(cnt[~is_pat] > 0).sum(axis=0),
    pat_reads=cnt[is_pat].sum(axis=0).astype(int), hc_reads=cnt[~is_pat].sum(axis=0).astype(int)))
per_gene.to_csv(OUT / 'marker_raw_per_gene.csv', index=False)
print('\nzero everywhere:', sorted(per_gene.loc[(per_gene.pat_reads == 0) & (per_gene.hc_reads == 0), 'symbol']))
print('top contributors:', per_gene.nlargest(4, 'pat_reads')[['symbol', 'pat_reads', 'hc_reads']].to_dict('records'))
```

```python
MARKER_MIN_DETECT = 1e-9  # tissue marker genes: any detection
DETECT_MIN = 0.10         # everything else

nz_frac = (cnt > 0).mean(axis=0)
marker_tbl['detect_frac'] = nz_frac
marker_tbl['detected'] = nz_frac >= MARKER_MIN_DETECT
marker_tbl.to_csv(OUT / 'marker_panel.csv', index=False)

mcols = np.array([gpos[e] for e in marker_tbl['ens']])
Zm_pat, Zm_hc = Z_neo[:, mcols], Z_hc_qc[:, mcols]            # matched arm (b1), DESeq2-paired
Zm_A, Zm_hc_A = Z_A[:, mcols], Z_hc_A[:, mcols]               # all arm (b1-b3), hit-level panels
zero_col = nz_frac == 0
print('randomized-quantile Z is not defined by data on a gene with no reads:')
print(f'  marker ENSG with zero counts in all {len(cnt)} samples: {int(zero_col.sum())}/{len(nz_frac)} '
      f'({sorted(marker_tbl.loc[zero_col, "symbol"])})')
print(f'  |Z|>{MARKER_Z} rate there: patient {(np.abs(Zm_pat[:, zero_col]) > MARKER_Z).mean():.3f}, '
      f'HC {(np.abs(Zm_hc[:, zero_col]) > MARKER_Z).mean():.3f}  (0.05 = pure noise)')
print(f'  on detected markers: patient {(np.abs(Zm_pat[:, marker_tbl["detected"]]) > MARKER_Z).mean():.3f}, '
      f'HC {(np.abs(Zm_hc[:, marker_tbl["detected"]]) > MARKER_Z).mean():.3f}')
print(f'\nany-hit rate on the full panel is a set-size counter, not a signal '
      f'(HC should sit at 1-0.95^n if every column were noise):')
for s in MARKER_SETS:
    c = marker_tbl['set'].values == s
    print(f'  {s:7s} n={int(c.sum()):2d}  noise expectation {1 - 0.95 ** c.sum():.2f}  '
          f'HC observed {(np.abs(Zm_hc[:, c]) > MARKER_Z).any(axis=1).mean():.2f}  '
          f'patient {(np.abs(Zm_pat[:, c]) > MARKER_Z).any(axis=1).mean():.2f}')
print(f'\nmarkers surviving detect_frac >= {MARKER_MIN_DETECT}: '
      f'{int(marker_tbl["detected"].sum())}/{len(marker_tbl)}')
print(marker_tbl[marker_tbl['detected']][['set', 'symbol', 'detect_frac']]
      .sort_values(['set', 'detect_frac'], ascending=[True, False]).round(3).to_string(index=False))
```

```python
# One record per arm: the samples a DESeq2 run was fitted on, their Z and hit matrices, the HC null pool
# (HC above the p99 of the full HC pool's n_sig are not null material), and that arm's DESeq2 results.
def build_arm(deg_dir, label):
    s = pd.read_csv(deg_dir / 'ruvg_W' / 'moore_b1__PancreaticNeoplasm.csv')['sample'].values
    st = pd.Series(s).map(stage_all).values
    case = np.isin(st, NEOPLASTIC)
    ip, ih = dis_idx.get_indexer(s[case]), hc_idx.get_indexer(s[~case])
    assert (ip >= 0).all() and (ih >= 0).all(), f'{label}: DESeq2 sample missing from the Z matrices'
    a = dict(label=label, Z=src['Z_dis'][ip], names=src['dis'][ip], stage=st[case],
             Z_hc=src['Z_hc'][ih], names_hc=src['hc'][ih])
    a['SIG'], a['SIG_HC'] = bh_sig_matrix(a['Z']), bh_sig_matrix(a['Z_hc'])
    a['SIG_M'], a['SIG_HC_M'] = marker_sig_matrix(a['Z'], names=a['names']), marker_sig_matrix(a['Z_hc'])
    null = pd.Series(a['names_hc']).map(nsig_of).values <= HC_NSIG_CUT
    a['Z_null'], a['SIG_NULL'], a['SIG_NULL_M'] = a['Z_hc'][null], a['SIG_HC'][null], a['SIG_HC_M'][null]
    a['deg'], a['padj'] = {}, {}
    for f in sorted(deg_dir.glob('moore_b1__*/*.csv.gz')):
        c, d = f.parent.name.split('__')[1], f.name.split('.')[0]
        r = pd.read_csv(f, index_col=0)
        a['deg'][(c, d)] = set(r.index[r['padj'].fillna(1.0) < DESEQ2_PADJ]) & gene_set
        a['padj'][(c, d)] = r[['padj', 'pvalue', 'log2FoldChange']]
    a['n_case'] = {c: int((~np.isin(pd.read_csv(deg_dir / 'ruvg_W' / f'moore_b1__{c}.csv')['sample'], src['hc'])).sum())
                   for c in {c for c, _ in a['deg']}}
    allres = pd.concat([v for (c, _), v in a['padj'].items() if c == 'PancreaticNeoplasm'])
    a['best_padj'] = allres.groupby(level=0)['padj'].min()
    a['best_lfc'] = allres.groupby(level=0)['log2FoldChange'].median()
    print(f"{label}: neoplasm {len(a['names'])} {dict(pd.Series(a['stage']).value_counts())}, "
          f"HC {len(a['names_hc'])} (null pool {int(null.sum())}), {len(a['deg'])} DESeq2 runs from {deg_dir.name}")
    return a


ARMS = {'A': build_arm(config.PANCREATIC_DEG_POOLED_DIR, 'Moore Batch_1-3'),
        'B': build_arm(config.PANCREATIC_DEG_DIR, 'Moore Batch_1')}
```

```python
from matplotlib.patches import Ellipse


# Genome-wide figures, one call per arm: Signal Number Heatmap (DESeq2 genes per case x design) and the
# UpSet of the four DESeq2 runs against the per-patient NM union, with the HC-union null as an inset.
ORDER = ['PDAC', 'IPMN', 'IsletCellTumor', 'PancreaticNeoplasm', 'Pancreatitis']
ORDER_LAB = ['PDAC', 'IPMN', 'Islet', 'Neoplasm', 'Pancreatitis']
DESIGN_ORDER = ['no_covariate', 'ruvg_k1', 'ruvg_k2', 'ruvg_k3']
UPSET_TOP = 12
VGROUPS = {'A': ['PDAC', 'IPMN', 'Islet Cell Tumor', 'all neoplasm'],
           'B': ['PDAC', 'IPMN', 'all neoplasm']}  # islet cell tumour is n=3 in the Batch_1 arm
DESIGN_NAME = {'no_covariate': 'DESeq2 Default', 'ruvg_k1': 'DESeq2 + RUVg (k=1)',
               'ruvg_k2': 'DESeq2 + RUVg (k=2)', 'ruvg_k3': 'DESeq2 + RUVg (k=3)'}
SET_COLOR = {DESIGN_NAME[d]: c for d, c in DESIGN_C.items()}
SET_COLOR['Normative Modeling (Union)'] = CPAL['red']


def arm_mask(a, grp):
    return np.ones(len(a['names']), bool) if grp == 'all neoplasm' else (a['stage'] == grp)


def arm_hc_draws(a, n, cols=None, M=None, k=N_HC_DRAWS, seed=SEED):
    M = a['SIG_NULL'] if M is None else M
    M = M if cols is None else M[:, cols]
    r = np.random.default_rng(seed)
    ns = min(n, M.shape[0])
    return ns, np.array([int(M[r.choice(M.shape[0], ns, replace=False)].any(axis=0).sum()) for _ in range(k)])


def signal_number_heatmap(a, p):
    t = (pd.DataFrame([dict(case=c, design=d, n=len(v)) for (c, d), v in a['deg'].items()])
         .pivot(index='case', columns='design', values='n').loc[ORDER, DESIGN_ORDER])
    t.to_csv(OUT / f'{p}_deseq2_nsig_by_case_design.csv')
    fig, ax = plt.subplots(figsize=(3.6, 2.12))
    ax.imshow(t.values, cmap=SEQ_CMAP, vmin=0, vmax=t.values.max(), aspect='auto')
    for i in range(t.shape[0]):
        for j in range(t.shape[1]):
            v = t.values[i, j]
            ax.text(j, i, v, ha='center', va='center', color='white' if v > t.values.max() * 0.55 else 'black')
    ax.set_xticks(range(4), ['none', 'k=1', 'k=2', 'k=3'])
    ax.set_yticks(range(5), [f'{lab} (n={a["n_case"][c]})' for lab, c in zip(ORDER_LAB, ORDER)])
    ax.set_xlabel('RUVg factors')
    ax.set_title(a['label'], loc='left', fontsize=FS_ANNOT_SM)
    for s in ax.spines.values():
        s.set_visible(False)
    fig.savefig(FIG / f'{p}_signal_number_heatmap.png', dpi=300, bbox_inches='tight')
    print(t.to_string())


def upset(ax_bar, ax_dot, sets, top=UPSET_TOP, ylabels=True):
    names = list(sets)
    cols = [SET_COLOR[n] for n in names]
    memb = {}
    for i, n in enumerate(names):
        for g in sets[n]:
            memb[g] = memb.get(g, 0) | (1 << i)
    combos = [c for c in Counter(memb.values()).most_common() if c[0]][:top]
    x = np.arange(len(combos))
    bar_col = [cols[int(np.log2(m))] if m and not m & (m - 1) else CPAL['dgrey'] for m, _ in combos]
    ax_bar.bar(x, [c[1] for c in combos], color=bar_col, width=0.65)
    for xi, (_, v) in zip(x, combos):
        ax_bar.text(xi, v, f'{v:,}', ha='center', va='bottom', rotation=90)
    ax_bar.set_xlim(-0.6, len(combos) - 0.4)
    ax_bar.set_xticks([])
    ax_bar.set_ylim(0, max(v for _, v in combos) * 1.35)
    ax_bar.set_axisbelow(True)
    ax_bar.grid(axis='y', color=CPAL_L['grey'], lw=0.8)
    for yi in range(len(names)):
        ax_dot.axhspan(yi - 0.5, yi + 0.5, color=cols[yi], alpha=0.10, lw=0)
    for xi, (mask, _) in zip(x, combos):
        on = [i for i in range(len(names)) if mask >> i & 1]
        ax_dot.plot([xi] * len(names), range(len(names)), 'o', ms=6, mfc='white', mec=CPAL_L['grey'], mew=0.8)
        if len(on) > 1:
            ax_dot.plot([xi, xi], [min(on), max(on)], '-', lw=1.2, color=CPAL['dgrey'], zorder=1)
        for i in on:
            ax_dot.plot([xi], [i], 'o', ms=6, mfc=cols[i], mec=cols[i], zorder=2)
    for i, n in enumerate(names):
        ax_dot.text(1.01, i, f'{len(sets[n]):,}', transform=ax_dot.get_yaxis_transform(),
                    ha='left', va='center', color=cols[i], fontsize=8)
    ax_dot.set_xlim(-0.6, len(combos) - 0.4)
    ax_dot.set_ylim(-0.6, len(names) - 0.4)
    ax_dot.set_xticks([])
    ax_dot.tick_params(axis='y', length=0)
    if ylabels:
        ax_dot.set_yticks(range(len(names)), names)
        for t, c in zip(ax_dot.get_yticklabels(), cols):
            t.set_color(c)
    else:
        ax_dot.set_yticks([])
    for sp in ax_dot.spines.values():
        sp.set_visible(False)


def null_inset(ax_bar, nmu_n, ns, draws):
    ax = ax_bar.inset_axes([0.62, 0.62, 0.46, 0.32])
    ax.barh([1, 0], [nmu_n, draws.mean()], 0.62, xerr=[0, draws.std()],
            color=[CPAL['red'], CPAL['dgrey']], error_kw=dict(lw=0.8, capsize=2))
    for y, v, c in [(1, nmu_n, CPAL['red']), (0, draws.mean(), CPAL['dgrey'])]:
        ax.text(v + nmu_n * 0.03, y, f'{v:,.0f}', va='center', fontsize=7, color=c)
    ax.set_yticks([1, 0], ['patients', f'null (HC, n={ns})'], fontsize=7)
    ax.set_xlim(0, nmu_n * 1.6)
    ax.set_xticks([])
    ax.tick_params(axis='y', length=0)
    ax.set_title(f'union size, {nmu_n / max(draws.mean(), 1):.0f}x null', fontsize=7.5, loc='left', pad=2)
    for sp in ax.spines.values():
        sp.set_visible(False)


def upset_summary(a, p):
    groups = VGROUPS[p]
    fig = plt.figure(figsize=(5.5 * len(groups), 4.6))
    gs = fig.add_gridspec(2, len(groups), height_ratios=[2.0, 1.1], hspace=0.05, wspace=0.26)
    rows = []
    for k, grp in enumerate(groups):
        m = arm_mask(a, grp)
        sets = {DESIGN_NAME[d]: a['deg'][(CASE_OF[grp], d)] for d in DESIGN_ORDER}
        sets['Normative Modeling (Union)'] = nmu = set(genes[a['SIG'][m].any(axis=0)])
        ns, draws = arm_hc_draws(a, int(m.sum()))
        ax_bar = fig.add_subplot(gs[0, k])
        upset(ax_bar, fig.add_subplot(gs[1, k], sharex=ax_bar), sets, ylabels=(k == 0))
        ax_bar.set_title(f'{grp}  (n={int(m.sum())})', loc='left')
        if k == 0:
            ax_bar.set_ylabel('genes')
        null_inset(ax_bar, len(nmu), ns, draws)
        f = plt.figure(figsize=(7.5, 4.6))
        g = f.add_gridspec(2, 1, height_ratios=[2.0, 1.1], hspace=0.05)
        ab = f.add_subplot(g[0, 0])
        upset(ab, f.add_subplot(g[1, 0], sharex=ab), sets)
        ab.set_ylabel('genes')
        ab.set_title(f'{grp}  (n={int(m.sum())})', loc='left')
        null_inset(ab, len(nmu), ns, draws)
        f.savefig(FIG / f"{p}_upset_{grp.replace(' ', '_')}.png", dpi=300, bbox_inches='tight')
        plt.close(f)
        rows.append(dict(group=grp, n_pat=int(m.sum()), nm_union=len(nmu),
                         group_deg_k2=len(a['deg'][(CASE_OF[grp], RUVG_PICK)]),
                         shared_k2=len(a['deg'][(CASE_OF[grp], RUVG_PICK)] & nmu),
                         hc_draw_mean=round(draws.mean(), 1), hc_draw_sd=round(draws.std(), 1)))
    fig.savefig(FIG / f'{p}_upset_summary.png', dpi=300, bbox_inches='tight')
    summary = pd.DataFrame(rows)
    summary.to_csv(OUT / f'{p}_upset_group_vs_individual.csv', index=False)
    print(summary.to_string(index=False))
```

```python
# Panel figures, one call per (arm, panel): cutoff sweep vs DESeq2, marker recovery (hit grid + bars),
# the full q-tier heatmap, and the individual-vs-group Venn. Tissue panels use the tissue-marker hit rule.
GRID_BLOCKS = 1
CELL_W, CELL_H = 0.26, 0.14  # inches per marker column / patient row
LABEL_IN = 1.5               # inches reserved under the heatmap for the rotated gene names
LABELS['grid_full_x'] = 'every marker called at q < 0.20: {n_markers} genes in {n_pat_hit} of {n_pat} patients'
LABELS['grid_full_y'] = '{n_pat_hit} patients'
VENN_TOP = 14
LINE_H = 0.20          # data units per text line at FS_ANNOT_SM, linespacing included
C_GRP = CPAL['grey']
PANEL_SPECS = {
    'pancreas': dict(tbl=marker_tbl, ylab='Pancreas marker genes called changed', title='pancreas cell-type markers (HPA)',
                     color=CPAL['red'], extra=['set'], detected_only=False, tissue=True,
                     row_colors=marker_tbl['set'].map(SET_C).values),
    'liver': dict(tbl=liver_tbl, ylab='Liver marker genes called changed', title='liver cell-type markers (HPA)',
                  color=LIVER_C, extra=['set'], detected_only=False, tissue=True, row_colors=None),
    'ot': dict(tbl=ot_tbl, ylab='OT pancreatic-cancer genes called changed', title='pancreatic-cancer genes (Open Targets)',
               color=CPAL['navy'], extra=['score'], detected_only=True, tissue=False, row_colors=None),
}


def arm_sig(a, key, q, hc=False, null=False):
    tissue = PANEL_SPECS[key]['tissue']
    if null:
        Z = a['Z_null']
        return marker_sig_matrix(Z, q=q) if tissue else bh_sig_matrix(Z, q=q)
    Z = a['Z_hc'] if hc else a['Z']
    if tissue:
        return marker_sig_matrix(Z, q=q, names=None if hc else a['names'])
    return bh_sig_matrix(Z, q=q)


def shades_of(color):
    return [mcolors.to_hex(1 - w + w * np.array(mcolors.to_rgb(color))) for w in (1.0, 0.66, 0.38)]


def marker_sweep(a, p, key):
    sp = PANEL_SPECS[key]
    mtbl, color = sp['tbl'], sp['color']
    cols = np.array([gpos[e] for e in mtbl['ens']])
    use = mtbl['detected'].values if sp['detected_only'] else np.ones(len(mtbl), bool)
    sig_q = {q: arm_sig(a, key, q)[:, cols] for q in Q_GRID}
    sig_hc_q = {q: arm_sig(a, key, q, null=True)[:, cols] for q in Q_GRID}
    n_pat, n_hcm = len(a['names']), len(a['Z_null'])
    ns = min(n_pat, n_hcm)
    rng_s = np.random.default_rng(SEED)
    hc_draws = {q: np.array([int(sig_hc_q[q][rng_s.choice(n_hcm, ns, replace=False)][:, use].any(axis=0).sum())
                             for _ in range(N_SWEEP_DRAWS)]) for q in Q_GRID}
    hc_union = {q: int(np.median(hc_draws[q])) for q in Q_GRID}
    ens = mtbl['ens'].values[use]
    deseq = {d: [int((a['padj'][('PancreaticNeoplasm', d)]['padj'].reindex(ens).fillna(1.0).values < q).sum())
                 for q in Q_GRID] for d in DESIGN_ORDER}
    nm = [int((sig_q[q][:, use].sum(axis=0) > 0).sum()) for q in Q_GRID]

    fig, ax = plt.subplots(figsize=(6.4, 6.4))
    for d in DESIGN_ORDER:
        ax.plot(Q_GRID, deseq[d], 'o-', color=SET_COLOR[DESIGN_NAME[d]], label=DESIGN_NAME[d], markersize=8, lw=1.5)
    ax.plot(Q_GRID, nm, 'o-', color=color, lw=2.5, label='Normative Modeling (Union)', markersize=8)
    ax.plot(Q_GRID, [hc_union[q] for q in Q_GRID], 'o--', color=CPAL['dgrey'], lw=1.5, markersize=8,
            label=f'Normative Null (HC, Union, n={ns} of {n_hcm})')
    ax.fill_between(Q_GRID, [int(hc_draws[q].min()) for q in Q_GRID], [int(hc_draws[q].max()) for q in Q_GRID],
                    color=CPAL['dgrey'], alpha=0.18, lw=0)
    ax.set_xscale('log')
    ax.set_xticks(Q_GRID, [str(q) for q in Q_GRID])
    ax.set_xlabel(LABELS['sweep_x'])
    ax.set_ylabel(LABELS['sweep_y'].format(ylab=sp['ylab'], n_panel=len(ens)))
    ax.set_title(a['label'], loc='left', fontsize=FS_ANNOT_SM)
    ax.grid(zorder=0, alpha=0.3)
    ax.legend(frameon=False, loc='upper left')
    fig.savefig(FIG / f'{p}_{key}_sweep.png', dpi=300, bbox_inches='tight')
    out = pd.DataFrame({'cutoff': Q_GRID, 'markers': nm,
                        'patients': [int((sig_q[q][:, use].sum(axis=1) > 0).sum()) for q in Q_GRID],
                        'hc_markers': [hc_union[q] for q in Q_GRID],
                        'hc_markers_max': [int(hc_draws[q].max()) for q in Q_GRID],
                        'excess_over_hc_max': [n - int(hc_draws[q].max()) for n, q in zip(nm, Q_GRID)],
                        'p_emp': [float((hc_draws[q] >= n).mean()) for n, q in zip(nm, Q_GRID)],
                        'n_pat': n_pat, 'n_hc_pool': n_hcm, 'n_hc_drawn': ns, 'n_draws': N_SWEEP_DRAWS, **deseq})
    out.to_csv(OUT / f'{p}_{key}_sweep.csv', index=False)
    print(out.to_string(index=False))


def marker_recovery(a, p, key):
    sp = PANEL_SPECS[key]
    mtbl, color, row_colors = sp['tbl'], sp['color'], sp['row_colors']
    stage = a['stage']
    cols = np.array([gpos[e] for e in mtbl['ens']])
    sig_A = {q: arm_sig(a, key, q)[:, cols] for q in Q_TIERS}
    use = mtbl['detected'].values if sp['detected_only'] else np.ones(len(mtbl), bool)
    tier = np.zeros(sig_A[Q_TIERS[0]].shape)
    for i, q in enumerate(sorted(Q_TIERS, reverse=True)):
        tier[sig_A[q]] = i + 1
    tier[:, ~use] = 0
    shades = shades_of(color)

    fig_grid, ax = plt.subplots(figsize=(11.0, 7.0))
    hit = tier > 0
    cand = list(np.where(hit.any(axis=0))[0])
    rows, shown = [], np.zeros(tier.shape[0], bool)
    while cand and len(rows) < TOP_ROWS:
        gain = [(int((hit[:, c] & ~shown).sum()), int(tier[:, c].max()), int(hit[:, c].sum())) for c in cand]
        if max(g[0] for g in gain) == 0:
            shown[:] = False
            continue
        best = cand[max(range(len(cand)), key=lambda i: gain[i])]
        rows.append(best)
        shown |= hit[:, best]
        cand.remove(best)
    rows = np.array(rows, dtype=int)
    rows = rows[np.lexsort((mtbl['symbol'].values[rows], -tier[:, rows].max(axis=0),
                            -(tier[:, rows] == len(Q_TIERS)).sum(axis=0)))]
    keep = np.where(tier.max(axis=1) > 0)[0]
    keep = keep[np.lexsort((-(tier[np.ix_(keep, rows)] == len(Q_TIERS)).sum(axis=1),
                            [list(STAGE_C).index(s) for s in stage[keep]]))]
    T = tier[np.ix_(keep, rows)].T
    ax.imshow(np.zeros(T.shape), cmap=SEQ_CMAP, vmin=0, vmax=1, aspect='auto')
    tier_h = []
    for lvl, sh in zip(range(len(Q_TIERS), 0, -1), shades):
        for y, x in zip(*np.where(T == lvl)):
            ax.add_patch(plt.Rectangle((x - 0.43, y - 0.43), 0.86, 0.86, color=sh, lw=0, zorder=2))
        tier_h.append(plt.Rectangle((0, 0), 1, 1, color=sh))
    for j, sm in enumerate(stage[keep]):
        ax.add_patch(plt.Rectangle((j - 0.5, -0.5), 1, len(rows), color=STAGE_C[sm], alpha=0.12, lw=0, zorder=0))
    ax.set_xlim(-0.5, len(keep) - 0.5)
    ax.set_ylim(len(rows) - 0.5, -0.5)
    ax.set_xticks(np.arange(-0.5, len(keep), 1), minor=True)
    ax.set_yticks(np.arange(-0.5, len(rows), 1), minor=True)
    ax.grid(which='minor', color='#FFFFFF', lw=0.8)
    ax.set_yticks(range(len(rows)), [mtbl['symbol'].values[r] for r in rows])
    if row_colors is not None:
        for t, r in zip(ax.get_yticklabels(), rows):
            t.set_color(row_colors[r])
    ax.set_xticks([])
    ax.set_xlabel(LABELS['grid_x'].format(n_pat_hit=len(keep), n_pat=tier.shape[0]))
    ax.set_ylabel(LABELS['grid_y'].format(n_rows=len(rows), n_markers_hit=int((tier.max(axis=0) > 0).sum())))
    ax.set_title(a['label'], loc='left', fontsize=FS_ANNOT_SM)
    ax.tick_params(length=0)
    h2 = [plt.Rectangle((0, 0), 1, 1, color=c, alpha=0.35) for c in STAGE_C.values()]
    l1 = [f'q < {q:.2f}' for q in sorted(Q_TIERS)]
    ax.legend([x for pair in zip(tier_h, h2) for x in pair], [x for pair in zip(l1, STAGE_C) for x in pair],
              frameon=False, loc='upper left', bbox_to_anchor=(0.1, -0.06), ncol=3,
              handletextpad=0.3, columnspacing=1.2, labelspacing=0.6)
    for s in ax.spines.values():
        s.set_visible(False)
    fig_grid.savefig(FIG / f'{p}_{key}_recovery_grid.png', dpi=300, bbox_inches='tight')

    fig_bar, ax = plt.subplots(figsize=(3, 6.4))
    pat_f = [(sig_A[q][:, use].sum(axis=1) > 0).mean() for q in sorted(Q_TIERS)]
    mrk_f = [(sig_A[q][:, use].sum(axis=0) > 0).mean() for q in sorted(Q_TIERS)]
    xs = [0, BAR_GAP]
    for x, (fr, n) in zip(xs, zip([pat_f, mrk_f], [tier.shape[0], int(use.sum())])):
        bot = 0
        for seg, sh in zip([fr[0], fr[1] - fr[0], fr[2] - fr[1]], shades):
            ax.bar(x, seg, bottom=bot, width=BAR_W, color=sh, lw=0)
            bot += seg
        ax.bar(x, 1 - bot, bottom=bot, width=BAR_W, color=CPAL_L['grey'], lw=0)
        ax.text(x, bot + 0.015, f'{bot:.0%}', ha='center', va='bottom', fontsize=12, fontweight='bold')
        ax.text(x, 1.04, LABELS['bar_n'].format(n=n), ha='center', va='bottom', fontsize=12)
    ax.set_xlim(-0.5 - BAR_W / 2, BAR_GAP + 0.5 + BAR_W / 2)
    ax.set_xticks(xs, list(LABELS['bar_x']))
    ax.set_ylim(0, 1.14)
    ax.set_yticks([0, 0.25, 0.5, 0.75, 1.0], ['0', '25%', '50%', '75%', '100%'])
    ax.tick_params(length=0)
    for s in ax.spines.values():
        s.set_visible(False)
    fig_bar.savefig(FIG / f'{p}_{key}_recovery_bars.png', dpi=300, bbox_inches='tight')
    out = pd.DataFrame({'cutoff': sorted(Q_TIERS),
                        'markers': [int((sig_A[q][:, use].sum(axis=0) > 0).sum()) for q in sorted(Q_TIERS)],
                        'patients': [int((sig_A[q][:, use].sum(axis=1) > 0).sum()) for q in sorted(Q_TIERS)],
                        'n_patients': tier.shape[0], 'n_panel': int(use.sum())})
    out.to_csv(OUT / f'{p}_{key}_recovery.csv', index=False)
    print(out.to_string(index=False))


def marker_heatmap(a, p, key):
    sp = PANEL_SPECS[key]
    mtbl, color, stage = sp['tbl'], sp['color'], a['stage']
    cols = np.array([gpos[e] for e in mtbl['ens']])
    use = mtbl['detected'].values
    sig_q = {q: arm_sig(a, key, q)[:, cols] & use[None, :] for q in Q_TIERS}
    tier = np.zeros(sig_q[Q_TIERS[0]].shape)
    for i, q in enumerate(sorted(Q_TIERS, reverse=True)):
        tier[sig_q[q]] = i + 1
    mk = np.where(tier.max(axis=0) > 0)[0]
    mk = mk[np.lexsort((mtbl['symbol'].values[mk], -(tier[:, mk] > 0).sum(axis=0), -tier[:, mk].max(axis=0)))]
    pt = np.where(tier[:, mk].max(axis=1) > 0)[0]
    pt = pt[np.lexsort((-(tier[np.ix_(pt, mk)] == len(Q_TIERS)).sum(axis=1),
                        [list(STAGE_C).index(s) for s in stage[pt]]))]
    shades = shades_of(color)
    grid_in = CELL_H * len(pt)
    fig, ax = plt.subplots(figsize=(CELL_W * len(mk) + 1.0, grid_in + LABEL_IN))
    T = tier[np.ix_(pt, mk)]
    ax.imshow(np.zeros(T.shape), cmap=SEQ_CMAP, vmin=0, vmax=1, aspect='auto')
    for lvl, sh in zip(range(len(Q_TIERS), 0, -1), shades):
        for y, x in zip(*np.where(T == lvl)):
            ax.add_patch(plt.Rectangle((x - 0.43, y - 0.43), 0.86, 0.86, color=sh, lw=0, zorder=2))
    for i, sm in enumerate(stage[pt]):
        ax.add_patch(plt.Rectangle((-0.5, i - 0.5), len(mk), 1, color=STAGE_C[sm], alpha=0.12, lw=0, zorder=0))
    ax.set_xlim(-0.5, len(mk) - 0.5)
    ax.set_ylim(len(pt) - 0.5, -0.5)
    ax.set_xticks(np.arange(-0.5, len(mk), 1), minor=True)
    ax.set_yticks(np.arange(-0.5, len(pt), 1), minor=True)
    ax.grid(which='minor', color='#FFFFFF', lw=0.8)
    ax.set_xticks(range(len(mk)), mtbl['symbol'].values[mk], rotation=90)
    ax.set_yticks([])
    ax.set_ylabel(LABELS['grid_full_y'].format(n_pat_hit=len(pt)))
    ax.set_title(a['label'], loc='left', fontsize=FS_ANNOT_SM)
    ax.tick_params(length=0)
    for s in ax.spines.values():
        s.set_visible(False)
    h1 = [plt.Rectangle((0, 0), 1, 1, color=sh) for sh in shades]
    h2 = [plt.Rectangle((0, 0), 1, 1, color=c, alpha=0.35) for c in STAGE_C.values()]
    ax.legend([x for pair in zip(h1, h2) for x in pair],
              [x for pair in zip([f'q < {q:.2f}' for q in sorted(Q_TIERS)], STAGE_C) for x in pair],
              frameon=False, loc='upper left', bbox_to_anchor=(0.35, -LABEL_IN / grid_in), ncol=3,
              handletextpad=0.3, columnspacing=1.2)
    ax.set_xlabel(LABELS['grid_full_x'].format(n_markers=len(mk), n_pat_hit=len(pt), n_pat=len(stage)))
    fig.savefig(FIG / f'{p}_{key}_heatmap.png', dpi=300, bbox_inches='tight')
    mtbl.iloc[mk].assign(**{f'n_pat_q{q:g}': sig_q[q][:, mk].sum(axis=0) for q in Q_TIERS}).to_csv(
        OUT / f'{p}_{key}_heatmap.csv', index=False)
    print(f'{p} {key} heatmap: {len(mk)} markers x {len(pt)} patients; '
          f'markers by cutoff { {q: int(sig_q[q].any(axis=0).sum()) for q in sorted(Q_TIERS)} }')


def case_table(a, key):
    sp = PANEL_SPECS[key]
    t = sp['tbl'].copy()
    c = np.array([gpos[e] for e in t['ens']])
    t['n_pat_bh'] = arm_sig(a, key, FDR_Q)[:, c].sum(axis=0)
    t['n_hc_bh'] = arm_sig(a, key, FDR_Q, hc=True)[:, c].sum(axis=0)
    t['group_padj'] = t['ens'].map(a['best_padj'])
    t['group_lfc'] = t['ens'].map(a['best_lfc'])
    t['indiv'] = t['n_pat_bh'] > 0
    t['group'] = t['group_padj'].fillna(1.0) < DESEQ2_PADJ
    t['measured'] = t['detect_frac'] > 0
    return t.sort_values(['n_pat_bh', 'group_padj'], ascending=[False, True])


def venn(a, p, key):
    sp = PANEL_SPECS[key]
    t = case_table(a, key)
    t.to_csv(OUT / f'{p}_{key}_case_table.csv', index=False)
    ok = t['measured']
    both, io = t[ok & t['indiv'] & t['group']], t[ok & t['indiv'] & ~t['group']]
    go = t[ok & ~t['indiv'] & t['group']].sort_values('group_padj')

    def block(df):
        s = df['symbol'].head(VENN_TOP).tolist()
        body = (s if s else ['(none)']) + ([f'+{len(df) - len(s)} more'] if len(df) > len(s) else [])
        return f'[{len(df)}]\n' + '\n'.join(body), len(body) + 1

    with plt.rc_context({k: FS_ANNOT_SM for k in ('font.size', 'axes.titlesize')}):
        fig, ax = plt.subplots(figsize=(6.5, 5.6))
        blocks = [block(d) for d in (io, both, go)]
        h = max(n for _, n in blocks) * LINE_H
        for x, c, lab, n in ((-1, sp['color'], 'Individual\nBH q<0.05, >=1 patient', len(both) + len(io)),
                             (1, C_GRP, 'Group\nDESeq2 padj<0.05\n(neoplasm vs HC, any design)', len(both) + len(go))):
            ax.add_patch(Ellipse((x, 0), 4.0, h + 1.0, facecolor=c, alpha=0.15, edgecolor=c, lw=1.2))
            ax.text(x * 1.9, h / 2 + 0.6, f'{lab}\nn={n}', ha='center', va='bottom', color=c)
        for x, (txt, _) in zip((-1.55, 0.0, 1.55), blocks):
            ax.text(x, h / 2, txt, ha='center', va='top', linespacing=1.4)
        ax.text(0, -h / 2 - 0.95, f'neither: {int((ok & ~t["indiv"] & ~t["group"]).sum())} genes silent on both sides',
                ha='center', va='top', color='0.35', fontsize=10)
        ax.set_title(f"{sp['title']} -- {a['label']}", fontsize=FS_ANNOT_SM + 2)
        ax.set_xlim(-3.3, 3.3)
        ax.set_ylim(-h / 2 - 1.7, h / 2 + 1.6)
        ax.set_aspect('equal')
        ax.axis('off')
        fig.savefig(FIG / f'{p}_{key}_venn.png', dpi=300, bbox_inches='tight')
    print(f'{p} {key}: individual {len(both) + len(io)}, group {len(both) + len(go)}, both {len(both)}; '
          f'also BH-significant in >=1 HC: {int((ok & (t["n_hc_bh"] > 0)).sum())}')


def panel_figures(p, key):
    a = ARMS[p]
    marker_sweep(a, p, key)
    marker_recovery(a, p, key)
    marker_heatmap(a, p, key)
    venn(a, p, key)
```

```python
# Severity gradient: per-sample deviation burden across the neoplastic series, with acute
# pancreatitis (non-neoplastic, same three batches) as the inflammatory contrast. All arm --
# no group contrast is involved, so every Moore Batch_1-3 sample is used.
from scipy.stats import kruskal, spearmanr

SEV_ORDER = NEOPLASTIC + ['Pancreatitis']
Z_panc = src['Z_dis'][a_panc]
sev = pd.concat([per_sample[['stage', 'n_sig']],
                 pd.DataFrame(dict(stage='Pancreatitis', n_sig=bh_sig_matrix(Z_panc).sum(axis=1)))])

neo_sev = sev[sev['stage'].isin(NEOPLASTIC)]
rho, p_rho = spearmanr(neo_sev['stage'].map({k: i for i, k in enumerate(NEOPLASTIC)}), neo_sev['n_sig'])
kw = kruskal(*[d['n_sig'].values for _, d in neo_sev.groupby('stage')]).pvalue
print(sev.groupby('stage')['n_sig'].agg(n='size', median='median', mean='mean')
      .reindex(SEV_ORDER).round(1).to_string())
print(f'neoplastic gradient: Spearman rho={rho:+.3f} p={p_rho:.2e} (n={len(neo_sev)}), '
      f'Kruskal p={kw:.2e}')
```

```python
def pair_jaccard(M, max_pairs=4000, seed=SEED):
    r = np.random.default_rng(seed)
    n = M.shape[0]
    pairs = [(i, j) for i in range(n) for j in range(i + 1, n)]
    if len(pairs) > max_pairs:
        pairs = [pairs[k] for k in r.choice(len(pairs), max_pairs, replace=False)]
    out = []
    for i, j in pairs:
        u = int((M[i] | M[j]).sum())
        if u:
            out.append(int((M[i] & M[j]).sum()) / u)
    return np.array(out)


def size_matched_null(M, seed=SEED):
    r = np.random.default_rng(seed)
    N = M.shape[1]
    Mn = np.zeros_like(M)
    for i, k in enumerate(M.sum(axis=1)):
        if k:
            Mn[i, r.choice(N, int(k), replace=False)] = True
    return Mn


j_pat = pair_jaccard(SIG)
j_hc = pair_jaccard(SIG_HC)
j_null = pair_jaccard(size_matched_null(SIG))
spec = pd.DataFrame([dict(set_='patients', n_pairs=len(j_pat), median=np.median(j_pat), mean=j_pat.mean()),
                     dict(set_='HC (batch-matched)', n_pairs=len(j_hc), median=np.median(j_hc), mean=j_hc.mean()),
                     dict(set_='size-matched null', n_pairs=len(j_null), median=np.median(j_null), mean=j_null.mean())])
spec.to_csv(OUT / 'patient_specificity_jaccard.csv', index=False)
print(spec.round(4).to_string(index=False))

share = pd.Series(carriers[carriers > 0]).value_counts().sort_index()
carrier_dist = share.rename('n_genes').rename_axis('n_patients_sharing').reset_index()
carrier_dist.to_csv(OUT / 'gene_carrier_distribution.csv', index=False)
print('\nhow many patients share a gene (the group-list premise is that this mass sits high)')
print(carrier_dist.head(10).to_string(index=False))
print(f"genes carried by exactly 1 patient: {int(share.get(1, 0))} of {int((carriers > 0).sum())} "
      f"({share.get(1, 0) / max((carriers > 0).sum(), 1):.0%})")
```

### 2. Genes found only by Moore: why NM did not call them
- The 18 genes with 0 BH-significant PDAC patients in NM, split by reason. Z is sign-aligned so that the Moore direction is up (+).
- **controls deviate (opposite)**: HC (benign) deviate from normal opposite to the Moore direction while PDAC is near normal. The group difference belongs to the controls.
- **cohort-wide shift**: HC also deviate in the same direction and PDAC goes only slightly further. A shift of the whole cohort.
- **small shift in PDAC**: HC are normal and PDAC overall shifts by < 1 SD. No patient crosses the individual threshold (the case NM misses).
- **sparse gene**: low detection rate; only a few samples have reads.
- Criteria: HC deviate if median HC Z exceeds ±`HC_OFFSET`; sparse if detection rate < `MIN_DETECT` (both arbitrary).
- Figure has the same structure as section 3:
  - Top = expression (log10 CPM), bottom = NM Z of the same samples.
  - Grey violin = full HC pool (NM reference), dark grey = this cohort's HC (benign), blue = PDAC, horizontal line = median.
  - On the Z axis: dotted line = uncorrected (|Z|>1.96), dashed line = BH threshold (median over PDAC samples).
- Genes are grouped by reason; label colour indicates the reason.

## Adding NM tissue-signal burden to the original study panel: Moore CEDAR -> BCC

- Moore 2025 selected 29 genes in CEDAR and validated them on BCC. That design is reproduced as is (literature panel CPM logistic, baseline), then the literature score is re-ranked by the **deviation count (burden)** of NM up-hits in a gene set.
- The Moore 29 were derived from CEDAR only, so with BCC as test there is no selection leakage on the literature-panel side.
- Claim: the literature panel already saturates train (train AUC ~1) and generalises poorly; the NM burden recovers test AUC.

| Arm | Burden genes | Labels used to pick genes |
|---|---|---|
| Baseline | Literature panel CPM (Moore 29 + nb6 pancreatic-cancer literature markers) | CEDAR only |
| T1 | All genes | none |
| T2 | NM tissue gene panel: nb6 tissue genes NM-detected in pancreatic cancer (`read10`: BH up in >= 1 patient with >= 10 reads, 0 same-batch HC) | **all 72 Moore patients incl. test -> leaky** |

- **Why re-ranking, not a joint model.** Putting literature CPM and burden in one logistic fails: both blocks almost perfectly separate train (train AUC 0.98 each, 0.998 combined), so there is no basis for learning weights, and CEDAR -> BCC test AUC drops to 0.720. So no weights are learned: each score is mapped to its empirical quantile in the **train distribution** and the two are summed, which also scores a single incoming sample.
- T2 is used as is, with the leak acknowledged: it shows how much the tissue signal NM found can add, not an unbiased test estimate.
- Dropped after testing (`_temp_fusion_repro.ipynb`): HPA liver panel and the 27 liver rising genes (severity ladder below) as burden (no gain; these genes rarely reach a BH hit) and as NM-Z features added to the literature logistic (test AUC fell).
- **Z source.** Disease and HC both use Z from the same CV fold model (produced by `cv_score_disease.py`). Mixing production Z (disease) with CV Z (HC) systematically inflates HC-side hits. Only this section uses this source; the sections above follow `SIG_SOURCE` as is.
- Per-(task, arm) scores are cached append-only in `fusion/tissue_burden.pkl` (delete a key to recompute); `fusion/tissue_burden.csv` is rebuilt from it.


```python
import scipy.sparse as sp
from scipy.stats import rankdata
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

FUSE_DIR = OUT / 'fusion'
FUSE_DIR.mkdir(parents=True, exist_ok=True)
HC_MAX_FRAC = 0.02    
MIN_DIS_HIT = 1      
N_BOOT = 2000
ZD_F = config.ZSCORES_DIR

meta_F = pd.concat([pd.read_csv(ZD_F / 'sample_meta.csv'),
                    pd.read_csv(ZD_F / 'hc_meta.csv').assign(phenotype='HC')], ignore_index=True)
meta_F['phenotype'] = meta_F['phenotype'].str.strip()
row_F = pd.Series(np.arange(len(meta_F)), index=meta_F['sample'])

Z_CACHE_F, G_CACHE_F = FUSE_DIR / 'z_cv.npy', FUSE_DIR / 'genes_cv.pkl'
if Z_CACHE_F.exists():
    Z_F = np.load(Z_CACHE_F)
    genes_F = pickle.load(open(G_CACHE_F, 'rb'))
else:
    folds_F = json.load(open(config.OUTRIDER_COMPARISON_DIR / 'cv_folds.json'))
    cv_order = pd.Index(np.concatenate([folds_F[k]['test'] for k in sorted(folds_F, key=int)]))
    hc_pos = cv_order.get_indexer(meta_F.loc[meta_F['phenotype'] == 'HC', 'sample'])
    assert (hc_pos >= 0).all()
    cv_sh = pickle.load(open(config.CV_DIR / 'cv_zscores_shash.pkl', 'rb'))
    keep_F = np.array([g in cv_sh and len(cv_sh[g]) == len(cv_order) and np.isfinite(cv_sh[g]).all() for g in genes])
    genes_F = genes[keep_F]
    Z_F = np.vstack([np.load(config.CV_DIR / 'Z_disease_cv_shash.npy')[:, keep_F],
                     np.column_stack([cv_sh[g][hc_pos] for g in genes_F])])
    Z_F = np.nan_to_num(np.clip(Z_F, -10, 10), nan=0.0).astype(np.float32)
    np.save(Z_CACHE_F, Z_F)
    pickle.dump(genes_F, open(G_CACHE_F, 'wb'))

H_CACHE_F = FUSE_DIR / 'bh_up_cv.npy'
if H_CACHE_F.exists():
    H_F = np.load(H_CACHE_F)
else:
    H_F = np.vstack([bh_fdr_reject(2 * norm.sf(np.abs(z)), q=FDR_Q) & (z > 0) for z in Z_F])
    np.save(H_CACHE_F, H_F)

C_F = sp.load_npz(config.DISEASE_TISSUE_DIR / 'all_counts.npz').tocsr()
assert C_F.shape == (len(meta_F), len(genes))
C_F = C_F.multiply(1e6 / np.asarray(C_F.sum(1))).tocsr()[:, pd.Index(genes).get_indexer(genes_F)]
CPM_F = np.log2(C_F.toarray() + 1).astype(np.float32)
del C_F
HC_ROWS_F = np.where(meta_F['phenotype'].eq('HC'))[0]
DET_F = (CPM_F > 0).mean(0)
print(f'genes {len(genes_F)}/{len(genes)} | Z {Z_F.shape} | HC pool {len(HC_ROWS_F)}')
```

```python
# literature panel = nb6 pancreatic-cancer literature markers + Moore 29 (CEDAR-derived)
lit_tbl = pd.read_csv(config.DISEASE_TISSUE_DIR / 'disease_tissue_genes.csv').query('lit')
lit_g = lit_tbl.loc[lit_tbl['phenotype'] == 'Pancreatic Cancer', 'gene'].drop_duplicates()
LIT_F = pd.Index(genes_F).get_indexer(lit_g)
m29_ens = pd.Series(MOORE29).map(sym2ensg).dropna()
m29_col = pd.Index([g.split('.')[0] for g in genes_F]).get_indexer(m29_ens)
LIT_F = np.union1d(LIT_F[LIT_F >= 0], m29_col[m29_col >= 0])
print(f'literature panel: {len(LIT_F)} genes (Moore 29 -> {(m29_col >= 0).sum()} mapped)')

moore_map = cmap[['sample', 'Cohort', 'Group']]
mo_F = meta_F.merge(moore_map, on='sample')
mo_F = mo_F[(mo_F['phenotype'] == 'HC') | mo_F['ood_keep'].eq(True)]
is_moore = meta_F['batch'].isin(MOORE_BATCHES)
panc = (meta_F['phenotype'] == 'Pancreatic Cancer') & meta_F['ood_keep'].eq(True)
lobo_rows = np.where(is_moore & (panc | meta_F['phenotype'].eq('HC')))[0]

TASKS_F = {
    'CEDAR -> BCC (PDAC vs Others)': dict(
        rows=row_F[mo_F['sample']].values, y=mo_F['Group'].eq('PDAC').astype(int).values,
        block=mo_F['Cohort'].values, scheme='fixed', train=['CEDAR'], test=['BCC']),
    'Moore LOBO (pancreatic cancer vs HC)': dict(
        rows=lobo_rows, y=panc.values[lobo_rows].astype(int),
        block=meta_F['batch'].values[lobo_rows], scheme='lobo'),
    'Moore CV (pancreatic cancer vs HC)': dict(
        rows=lobo_rows, y=panc.values[lobo_rows].astype(int),
        block=meta_F['batch'].values[lobo_rows], scheme='cv'),
}
for k, t in TASKS_F.items():
    print(f'{k}: n={len(t["rows"])}, patients {int(t["y"].sum())}, block {sorted(set(t["block"]))}')
```

```python
def splits_F(t):
    y, b = t['y'], t['block']
    if t['scheme'] == 'fixed':
        yield np.where(np.isin(b, t['train']))[0], np.where(np.isin(b, t['test']))[0]
    elif t['scheme'] == 'cv':
        yield from StratifiedKFold(5, shuffle=True, random_state=SEED).split(y, y)
    else:
        for tb in sorted(set(b)):
            te = np.where(b == tb)[0]
            if len(set(y[te])) == 2:
                yield np.where(b != tb)[0], te


def _fit(X, y):
    return make_pipeline(StandardScaler(), LogisticRegression(C=1.0, class_weight='balanced',
                                                              max_iter=2000)).fit(X, y)


def fuse_tt(rows_tr, rows_te, y_tr, cols):
    clf = _fit(CPM_F[np.ix_(rows_tr, LIT_F)], y_tr)
    s_tr, s_te = (clf.decision_function(CPM_F[np.ix_(r, LIT_F)]) for r in (rows_tr, rows_te))
    b_tr, b_te = (H_F[np.ix_(r, cols)].sum(1).astype(float) for r in (rows_tr, rows_te))
    return dict(lit=(s_tr, s_te), burden=(b_tr, b_te), fuse=(qf(s_tr, s_tr) + qf(b_tr, b_tr), qf(s_tr, s_te) + qf(b_tr, b_te)))


tis_g = (pd.read_csv(config.DISEASE_TISSUE_DIR / 'disease_tissue_genes.csv')
         .query("phenotype == 'Pancreatic Cancer' and rule == 'read10' and nm_up")['gene'].unique())
TISSUE_F = pd.Index(genes_F).get_indexer(tis_g)
TISSUE_F = np.unique(TISSUE_F[TISSUE_F >= 0])
TIS_ARMS = [('T1. all genes', lambda t: np.arange(len(genes_F))),
            ('T2. NM tissue gene panel (leaky)', lambda t: TISSUE_F)]
KEYS = ('lit', 'burden', 'fuse')
qf = lambda ref, v: np.searchsorted(np.sort(ref), v, side='right') / (len(ref) + 1)
TIS_PKL = FUSE_DIR / 'tissue_burden.pkl'
print(f'NM tissue gene panel {len(TISSUE_F)}/{len(tis_g)}')

tis_res = pickle.load(open(TIS_PKL, 'rb')) if TIS_PKL.exists() else {}
for tname, t in TASKS_F.items():
    for lab, cols_of in TIS_ARMS:
        if (tname, lab) in tis_res:
            continue
        cols = cols_of(t)
        r = dict(n_panel=[], y_tr=[], tr={k: [] for k in KEYS}, te={k: [] for k in KEYS})
        for tr, tst in splits_F(t):
            r['n_panel'].append(len(cols))
            r['y_tr'].append(t['y'][tr])
            for k, (a, b) in fuse_tt(t['rows'][tr], t['rows'][tst], t['y'][tr], cols).items():
                r['tr'][k].append(a)
                r['te'][k].append(b)
        r['te'] = {k: np.concatenate(v) for k, v in r['te'].items()}
        tis_res[(tname, lab)] = r
        pickle.dump(tis_res, open(TIS_PKL, 'wb'))
        print(f'{tname} | {lab}: done')
```

```python
def auc_fast(y, s):
    r = rankdata(s)
    n1 = y.sum()
    return (r[y == 1].sum() - n1 * (n1 + 1) / 2) / (n1 * (len(y) - n1))


LIT_LAB = '1. Literature + CPM'
TIS_ORDER = [LIT_LAB] + [a for a, _ in TIS_ARMS]
y_pool_F, test_rows_F, BOOT_IDX, boot_te, boot_tr, rows_t = {}, {}, {}, {}, {}, []
for tname, t in TASKS_F.items():
    y0 = y_pool_F[tname] = np.concatenate([t['y'][te] for _, te in splits_F(t)])
    test_rows_F[tname] = np.concatenate([t['rows'][te] for _, te in splits_F(t)])
    rng = np.random.default_rng(SEED)
    BOOT_IDX[tname] = [i for i in (rng.choice(len(y0), len(y0), replace=True) for _ in range(N_BOOT))
                       if len(set(y0[i])) == 2]
    r0 = tis_res[(tname, TIS_ARMS[0][0])]
    rng = np.random.default_rng(SEED)
    tr_idx = [[np.concatenate([rng.choice(np.where(y == c)[0], int((y == c).sum())) for c in (0, 1)])
               for _ in range(N_BOOT)] for y in r0['y_tr']]
    boot_tr_of = lambda sc: np.mean([[auc_fast(y[i], s[i]) for i in idx]
                                     for y, s, idx in zip(r0['y_tr'], sc, tr_idx)], axis=0)
    boot_te[(tname, LIT_LAB)] = np.array([auc_fast(y0[i], r0['te']['lit'][i]) for i in BOOT_IDX[tname]])
    boot_tr[(tname, LIT_LAB)] = boot_tr_of(r0['tr']['lit'])
    for lab, _ in TIS_ARMS:
        r = tis_res[(tname, lab)]
        boot_te[(tname, lab)] = np.array([auc_fast(y0[i], r['te']['fuse'][i]) for i in BOOT_IDX[tname]])
        boot_tr[(tname, lab)] = boot_tr_of(r['tr']['fuse'])
    for lab in TIS_ORDER:
        d = boot_te[(tname, lab)] - boot_te[(tname, LIT_LAB)]
        r = tis_res[(tname, TIS_ARMS[0][0] if lab == LIT_LAB else lab)]
        key = 'lit' if lab == LIT_LAB else 'fuse'
        rows_t.append(dict(task=tname, arm=lab, n_panel=None if lab == LIT_LAB else int(np.median(r['n_panel'])),
                           train_auc=np.mean([auc_fast(y, s) for y, s in zip(r['y_tr'], r['tr'][key])]),
                           test_auc=roc_auc_score(y0, r['te'][key]),
                           added_alone_test=None if lab == LIT_LAB else roc_auc_score(y0, r['te']['burden']),
                           delta_mean=d.mean(), lo=np.percentile(d, 2.5), hi=np.percentile(d, 97.5),
                           p_le0=float((d <= 0).mean())))
tis_tab = pd.DataFrame(rows_t).round(3)
tis_tab.to_csv(FUSE_DIR / 'tissue_burden.csv', index=False)
tis_tab
```

### Individual patients changed by the burden re-ranking (BCC, all reported)

- Compares each sample's **percentile relative to controls** between M0 (Moore marker + CPM) and M2 (+ NM tissue gene panel burden, T2).
- A single threshold only shows how many crossed it, so all 17 patients and the changes on the non-PDAC side are also shown.
- `pct_M0` / `pct_M2`: fraction of non-PDAC samples in the same test set scoring below that sample.
- `M0@90` / `M2@90`: whether it crosses the threshold at 90% specificity (non-PDAC reference).
- `burden`: number of NM tissue gene panel genes with BH-significant up-deviation; `top_genes`: top of those by Z.
- The table is saved to `fusion/bcc_case_table.csv`.

## Severity ladder: which deviations rise with lesion type and PDAC stage, and where do they come from?

Moore et al. 2025 only ran the binary PDAC vs remaining-patients contrast, which cannot show the
**ordinal structure** across lesion types and stages. The search here is genome-wide rather than
restricted to a tissue panel, and the tissue question is asked afterwards, of the genes that were found.

Samples are the Moore Batch_1-3 series, split into eight blocks:
Benign control (= the Moore benign controls, all of them in the NM HC pool) < Pancreatitis < IPMN <
Islet Cell Tumor < PDAC stage I, II, III, IV. `Cancer_other` is left out, its lesion is not specified.

A gene is called **rising** when all three hold:
1. lesion ladder (IPMN < Islet < PDAC) Spearman `rho > 0` at nominal `p < 0.05`,
2. PDAC stage ladder (I < II < III < IV) `rho > 0`,
3. quiet in the non-neoplastic blocks: 0 of 71 benign controls and at most 1 of 79 pancreatitis
   patients carry a BH-significant up-hit.

No detection-rate filter: a gene detected in few samples is exactly the tail signal this project is
after, and RQR noise at zero counts is already blocked by the hit rule (>= 10 reads).
BH over 19,917 genes is unreachable here (the smallest p is 2.6e-6 against a 2.5e-6 threshold), so
selection uses the nominal p and BH is reported as a column only.

**What this supports.** The figures below are descriptive: the genes were chosen with the same labels
the trends are then read against, so the correlation coefficients measure the strength of that
selection as much as the biology. The claim they carry is the one about composition -- *where* the
rising genes come from -- not the size of any single correlation.

```python
# Severity blocks and the genome-wide search for rising genes.
from scipy.stats import rankdata, spearmanr, t as t_dist

LAD_DIR = OUT / 'severity_ladder'
LAD_DIR.mkdir(parents=True, exist_ok=True)
LAD_ORDER = ['Healthy Control', 'Pancreatitis', 'IPMN', 'Islet Cell Tumor', 'PDAC']
BLK_LAB = ['Benign control', 'Pancreatitis', 'IPMN', 'Islet cell tumour',
           'PDAC I', 'PDAC II', 'PDAC III', 'PDAC IV']
QUIET_BENIGN, QUIET_PANC, SEL_P, LAD_Q = 0, 1, 0.05, 0.05
FSP = {k: FS_ANNOT_LG for k in ('font.size', 'axes.labelsize', 'xtick.labelsize',
                                'ytick.labelsize', 'legend.fontsize')}

names_L = np.concatenate([names_hc_A, src['dis'][a_panc], names_A])
stage_L = np.concatenate([np.repeat('Healthy Control', len(names_hc_A)),
                          np.repeat('Pancreatitis', int(a_panc.sum())), stage_A])
Z_L = np.vstack([Z_hc_A, src['Z_dis'][a_panc], Z_A])
rank_L = pd.Series(stage_L).map({s: i for i, s in enumerate(LAD_ORDER)}).values
ts_L = pd.to_numeric(pd.Series(names_L).map(meta['Tumor_Stage']), errors='coerce').values
is_pdac, neo_L = stage_L == 'PDAC', rank_L >= 2
blk_L = np.where(is_pdac, 3 + ts_L, rank_L).astype(int)
assert not pd.isna(rank_L).any() and np.isfinite(ts_L[is_pdac]).all()
HIT_L = marker_sig_matrix(Z_L, names=names_L)

# Vectorised Spearman: rank once, then Pearson. Same tie handling and same t approximation as scipy,
# but all 19,917 genes in one pass instead of a per-gene call.
def rank_z(M):
    R = rankdata(M, axis=0)
    R = R - R.mean(0)
    sd = R.std(0)
    return R / np.where(sd == 0, 1, sd)

def rank_x(x):
    r = rankdata(x)
    return (r - r.mean()) / r.std()

def rho_p(rx, RZ):
    rho = (rx @ RZ) / len(rx)
    with np.errstate(divide='ignore', invalid='ignore'):
        tt = rho * np.sqrt((len(rx) - 2) / np.clip(1 - rho ** 2, 1e-12, None))
    return rho, 2 * t_dist.sf(np.abs(tt), len(rx) - 2)

x_neo, x_pd = rank_L[neo_L].astype(float), ts_L[is_pdac]
LAD_PATH = LAD_DIR / 'ladder_all_genes.csv'
if LAD_PATH.exists():
    lad = pd.read_csv(LAD_PATH)
else:
    rho_neo, p_neo = rho_p(rank_x(x_neo), rank_z(Z_L[neo_L]))
    rho_st, p_st = rho_p(rank_x(x_pd), rank_z(Z_L[is_pdac]))
    lad = pd.DataFrame(dict(ens=genes, rho_neo=rho_neo, p_neo=p_neo, rho_stage=rho_st, p_stage=p_st))
    for i, lab in enumerate(BLK_LAB):
        m = blk_L == i
        lad[f'medZ_{lab}'] = np.median(Z_L[m], axis=0)
        lad[f'hit_{lab}'] = HIT_L[m].mean(axis=0)
    lad.to_csv(LAD_PATH, index=False)

# Quiet is counted in patients, not as a rate: a 5% rate lets 99% of the transcriptome through.
n_ben, n_pan = int((blk_L == 0).sum()), int((blk_L == 1).sum())
lad['quiet'] = ((lad['hit_Benign control'] * n_ben <= QUIET_BENIGN) &
                (lad['hit_Pancreatitis'] * n_pan <= QUIET_PANC))
lad['bh_neo'] = bh_fdr_reject(lad['p_neo'].values, q=LAD_Q)
lad['score'] = lad[['rho_neo', 'rho_stage']].mean(axis=1)
lad['pool'] = (lad['rho_neo'] > 0) & (lad['rho_stage'] > 0) & lad['quiet'] & (lad['p_neo'] < SEL_P)

print(pd.Series(BLK_LAB)[blk_L].value_counts().reindex(BLK_LAB).to_frame('n').T.to_string())
print(f'\ntranscriptome {len(lad)} genes')
print(f"  both rho > 0: {int(((lad.rho_neo > 0) & (lad.rho_stage > 0)).sum())}")
print(f"  + benign/pancreatitis quiet: {int(((lad.rho_neo > 0) & (lad.rho_stage > 0) & lad.quiet).sum())}")
print(f"  + lesion ladder p < {SEL_P} (= rising): {int(lad['pool'].sum())}")
print(f"  BH q < {LAD_Q} for reference: {int(lad['bh_neo'].sum())} "
      f"(smallest p_neo {lad['p_neo'].min():.1e} vs threshold {LAD_Q / len(lad):.1e})")
print(f'-> {LAD_PATH}')
```

```python
# Each tissue against the whole selected set. Per-gene mean Z is the fair unit -- the liver panel
# holds the most rising genes, so a raw Z sum would favour it by size alone.
MIN_GENES_T = 3
ALL_LAB = f'all rising genes ({len(pool)})'
tis = [p for p, n in pool['panel'].value_counts().items() if n >= MIN_GENES_T]
col_of = lambda g: pd.Index(genes).get_indexer(g['ens'])
Zmean = {p: Z_L[:, col_of(pool[pool['panel'] == p])].mean(1) for p in tis}
Zmean[ALL_LAB] = Z_L[:, col_of(pool)].mean(1)
Zmean['unassigned'] = Z_L[:, col_of(pool[pool['panel'].isna()])].mean(1)

rows_t = []
for p, v_ in Zmean.items():
    n = len(pool) if p == ALL_LAB else int(pool['panel'].isna().sum()) if p == 'unassigned' \
        else int((pool['panel'] == p).sum())
    base = np.median(v_[blk_L == 0])
    rows_t.append(dict(panel=p, n_genes=n, rise=np.median(v_[blk_L == 7]) - base,
                       rise_sum=(np.median(v_[blk_L == 7]) - base) * n,
                       step=np.median(v_[blk_L >= 4]) - np.median(v_[blk_L < 4]),
                       rho_neo=spearmanr(rank_L[neo_L], v_[neo_L]).statistic,
                       rho_stage=spearmanr(ts_L[is_pdac], v_[is_pdac]).statistic,
                       median_benign=base, median_pdac4=np.median(v_[blk_L == 7])))
trend_t = pd.DataFrame(rows_t).sort_values('rise', ascending=False).reset_index(drop=True)
trend_t.to_csv(LAD_DIR / 'tissue_trend.csv', index=False)

n_ann_pool = int(pool['panel'].notna().sum())
n_liver_pool = int((pool['panel'] == 'liver').sum())
n_ann_all = int(lad['panel'].notna().sum())
n_liver_all = int((lad['panel'] == 'liver').sum())
ref_rise = float(trend_t.loc[trend_t['panel'] == ALL_LAB, 'rise'].iloc[0])

# a white halo under each highlighted line keeps it readable where tissues cross
HALO = [pe.Stroke(linewidth=4.2, foreground='white'), pe.Normal()]
HL_LAST = 'macrophage'   # highlight the ranked run from the top tissue down to this one
REST_C, ALL_C = CPAL_L['grey'], CPAL['dgrey']
rank_t = [p for p in trend_t['panel'] if p not in (ALL_LAB, 'unassigned')]
hl_t = rank_t[:rank_t.index(HL_LAST) + 1] if HL_LAST in rank_t else rank_t
# same tissue palette as the heatmap strip above, so a tissue keeps its colour across both figures
panel_c = lambda p: ALL_C if p == ALL_LAB else P_C.get(p, REST_C) if p in hl_t else REST_C

with plt.rc_context(FSP):
    fig, (axl, axr) = plt.subplots(1, 2, figsize=(14, 5.6), gridspec_kw=dict(width_ratios=[1.6, 1], wspace=0.32))
    # drawn worst-ranked first so the top tissue sits on top of every crossing line
    ends = []
    for p in trend_t['panel'][::-1]:
        v_ = Zmean[p]
        med_ = [np.median(v_[blk_L == b]) for b in range(8)]
        if p == ALL_LAB:
            axl.plot(range(8), med_, lw=2.4, ls='--', color=ALL_C, zorder=20, path_effects=HALO)
            ends.append([med_[7], ALL_LAB, ALL_C])
        elif p in hl_t:
            k = hl_t.index(p)
            axl.plot(range(8), med_, lw=2.4, marker='o', ms=5, color=panel_c(p), zorder=30 - k,
                     solid_capstyle='round', path_effects=HALO)
            ends.append([med_[7], f"{p} ({int(trend_t.loc[trend_t['panel'] == p, 'n_genes'].iloc[0])})", panel_c(p)])
        else:
            axl.plot(range(8), med_, lw=0.9, color=REST_C, zorder=1, alpha=0.6)
    gap = 0.115 * (max(e[0] for e in ends) - min(e[0] for e in ends) + 1e-9)
    ends.sort(key=lambda e: -e[0])
    for i in range(1, len(ends)):
        ends[i][0] = min(ends[i][0], ends[i - 1][0] - gap)
    for y_lab, name, c_ in ends:
        axl.annotate(name, (7.12, y_lab), va='center', ha='left', fontsize=FS_ANNOT_SM, color=c_)
    axl.axhline(0, color='0.3', lw=1.0, ls=':')
    axl.axvline(3.5, color='0.55', lw=1.4)
    axl.text(3.42, axl.get_ylim()[1], 'other lesions  ', ha='right', va='top', fontsize=FS_ANNOT_SM, color='0.45')
    axl.text(3.58, axl.get_ylim()[1], '  PDAC', ha='left', va='top', fontsize=FS_ANNOT_SM, color='0.45')
    axl.set_xticks(range(8))
    axl.set_xticklabels([f'{l} (n={int((blk_L == b).sum())})' for b, l in enumerate(BLK_LAB)],
                        rotation=35, ha='right', rotation_mode='anchor')
    axl.set_ylabel('Median deviation Z per gene')
    axl.set_xlim(-0.3, 9.2)
    axl.grid(axis='y', color='0.92', lw=0.8)
    axl.set_axisbelow(True)
    axl.set_title('Deviation per gene, by tissue')

    y = np.arange(len(trend_t))[::-1]
    cols = [panel_c(p) for p in trend_t['panel']]
    axr.barh(y, trend_t['rise'], color=cols, edgecolor='black', lw=0.6, height=0.68)
    axr.axvline(ref_rise, color=ALL_C, lw=1.6, ls='--', zorder=4)
    for i, r in trend_t.iterrows():
        axr.text(r['rise'] + trend_t['rise'].max() * 0.03, y[i],
                 f"{r['rise']:+.2f}   (summed {r['rise_sum']:+.1f})", va='center', ha='left',
                 fontsize=FS_ANNOT_SM)
    axr.set_yticks(y)
    axr.set_yticklabels([p if p == ALL_LAB else f"{p} ({n})"
                         for p, n in zip(trend_t['panel'], trend_t['n_genes'])])
    axr.set_xlabel('Rise per gene  (benign median -> PDAC IV median)')
    axr.set_xlim(0, trend_t['rise'].max() * 1.9)
    axr.grid(axis='x', color='0.92', lw=0.8)
    axr.set_axisbelow(True)
    axr.set_title('Rise per gene, ranked')
    fig.suptitle(f'Liver markers carry the rise: {n_liver_pool}/{n_ann_pool} '
                 f'({n_liver_pool / n_ann_pool:.0%}) of tissue-annotated rising genes are liver, '
                 f'vs {n_liver_all / n_ann_all:.0%} of annotated genes overall', y=1.02)
    fig.savefig(FIG / 'ladder_tissue_trend.png', dpi=300, bbox_inches='tight')
    plt.show()
display(trend_t.round(3))
print(f"-> {FIG / 'ladder_tissue_trend.png'}")
```

```python
# Per-gene trajectories and the per-sample panel score, for the whole set and for liver alone.
def trend_fig(gdf, label, fname):
    cols = pd.Index(genes).get_indexer(gdf['ens'])
    med_ = np.vstack([np.median(Z_L[blk_L == b][:, cols], axis=0) for b in range(8)])
    sc = Z_L[:, cols].mean(axis=1)
    r_all = spearmanr(blk_L[blk_L >= 2], sc[blk_L >= 2])
    r_pd = spearmanr(ts_L[is_pdac], sc[is_pdac])
    xt = [f'{l} (n={int((blk_L == b).sum())})' for b, l in enumerate(BLK_LAB)]

    with plt.rc_context(FSP):
        fig, (axl, axr) = plt.subplots(1, 2, figsize=(13, 5.0), gridspec_kw=dict(wspace=0.26))
        for j in range(med_.shape[1]):
            axl.plot(range(8), med_[:, j], color='0.72', lw=0.7, alpha=0.7)
        axl.plot(range(8), med_.mean(axis=1), color='black', lw=3.0, marker='o', ms=7,
                 label=f'mean of {med_.shape[1]} genes')
        axl.axhline(0, color='0.3', lw=1.0, ls=':')
        axl.set_ylabel('Median deviation Z')
        axl.legend(frameon=False, loc='upper left')

        bp = axr.boxplot([sc[blk_L == b] for b in range(8)], positions=range(8), widths=0.62,
                         patch_artist=True, showfliers=False, medianprops=dict(color='black', lw=2.6),
                         whiskerprops=dict(lw=1.6), capprops=dict(lw=1.6),
                         boxprops=dict(lw=1.6, edgecolor='black'))
        for b, box in enumerate(bp['boxes']):
            box.set_facecolor(BLK_C8[b])
            box.set_alpha(0.9)
        rngp = np.random.default_rng(SEED)
        for b in range(8):
            vv = sc[blk_L == b]
            axr.plot(b + rngp.normal(0, 0.07, len(vv)), vv, '.', ms=4.5, color='0.25', alpha=0.6)
        axr.axhline(0, color='0.3', lw=1.0, ls=':')
        axr.set_ylabel('Panel score (mean Z over the genes)')
        axr.text(0.03, 0.97, f'IPMN -> PDAC IV:  rho={r_all.statistic:+.2f}, p={r_all.pvalue:.1e}\n'
                             f'within PDAC:       rho={r_pd.statistic:+.2f}, p={r_pd.pvalue:.1e}',
                 transform=axr.transAxes, va='top', ha='left', fontsize=FS_ANNOT_LG - 2, linespacing=1.5)
        for a in (axl, axr):
            a.axvline(3.5, color='0.55', lw=1.4)
            a.text(3.42, a.get_ylim()[1], 'other lesions  ', ha='right', va='top',
                   fontsize=FS_ANNOT_SM, color='0.45')
            a.text(3.58, a.get_ylim()[1], '  PDAC', ha='left', va='top',
                   fontsize=FS_ANNOT_SM, color='0.45')
            a.set_xticks(range(8))
            a.set_xticklabels(xt, rotation=35, ha='right', rotation_mode='anchor')
            a.set_xlim(-0.6, 7.6)
            a.grid(axis='y', color='0.92', lw=0.8)
            a.set_axisbelow(True)
        fig.suptitle(f'{label} deviation rises with lesion type and PDAC stage', y=1.02)
        fig.savefig(FIG / fname, dpi=300, bbox_inches='tight')
        plt.show()
    print(f'{label}: {len(gdf)} genes, IPMN->PDAC IV rho {r_all.statistic:+.3f} (p {r_all.pvalue:.1e}), '
          f'within PDAC rho {r_pd.statistic:+.3f} (p {r_pd.pvalue:.1e})  -> {FIG / fname}')

trend_fig(pool, f'Genome-wide rising genes ({len(pool)})', 'ladder_trend_all.png')
trend_fig(pool[pool['panel'] == 'liver'], 'Liver marker rising genes', 'ladder_trend_liver.png')
```

### Is the liver signal just liver metastasis, or just CA19-9?

The liver panel carries the largest share of the rising genes, so the obvious objection is that it
restates something already measured in the clinic. Two tests inside the PDAC patients, both on the
per-sample liver panel score (sum of Z over the liver genes):

- **liver metastasis** as a group split (Mann-Whitney), using the `Met_Site` field,
- **CA19-9** as a rank correlation (Spearman), on the patients who have a value.

Both are negative, which is the point: the axis is weak but it is not a restatement of either.
Note the asymmetry of the split -- liver metastasis is recorded only in stage IV, so the contrast is
7 against 30 and cannot exclude a modest effect.


```python
# Does the liver panel score restate liver metastasis or CA19-9? Two simple tests inside PDAC.
from scipy.stats import mannwhitneyu, theilslopes

C_MET3, C_NOMET2, C_CA = CPAL['orange'], CPAL['blue'], CPAL['purple']
with h5py.File(config.H5AD_PATH, 'r') as h:
    idx_o = pd.Index(dec(h['obs/_index'][()]))
    def obs_col(n):
        o = h[f'obs/{n}']
        return pd.Series(pd.Categorical.from_codes(o['codes'][()], dec(o['categories'][()])).astype(str)
                         if isinstance(o, h5py.Group) else np.asarray(o[()]), index=idx_o)
    met_all, ca_all = obs_col('Met_Site'), obs_col('CA19_9')

liver_met = np.array(['liver' in m for m in
                      pd.Series(names_L).map(met_all).fillna('not recorded').str.lower().values])
lca = np.log10(np.clip(pd.to_numeric(pd.Series(names_L).map(ca_all), errors='coerce').values, 0, None) + 1)
ok_ca = np.isfinite(lca)
liver_cols = pd.Index(genes).get_indexer(pool.loc[pool['panel'] == 'liver', 'ens'])
v_liv = Z_L[:, liver_cols].sum(1)

g_no, g_met = is_pdac & ~liver_met, is_pdac & liver_met
u_met = mannwhitneyu(v_liv[g_met], v_liv[g_no])
m_ca = ok_ca & is_pdac
r_ca = spearmanr(lca[m_ca], v_liv[m_ca])
stars = lambda p: 'n.s.' if p >= 0.05 else ('*' if p >= 0.01 else ('**' if p >= 0.001 else '***'))

pd.DataFrame([dict(test='liver met vs none (Mann-Whitney)', n1=int(g_met.sum()), n2=int(g_no.sum()),
                   stat=u_met.statistic, p=u_met.pvalue,
                   median1=np.median(v_liv[g_met]), median2=np.median(v_liv[g_no])),
              dict(test='liver score vs log10 CA19-9 (Spearman)', n1=int(m_ca.sum()), n2=np.nan,
                   stat=r_ca.statistic, p=r_ca.pvalue, median1=np.nan, median2=np.nan)]
             ).to_csv(LAD_DIR / 'liver_independence.csv', index=False)

YL = f'Liver panel score (sum Z, {len(liver_cols)} genes)'
GRP = [('no liver\nmetastasis', g_no, C_NOMET2), ('liver\nmetastasis', g_met, C_MET3)]
rngs = np.random.default_rng(0)
with plt.rc_context(FSP):
    fig, (axm, axc) = plt.subplots(1, 2, figsize=(9.5, 4.4),
                                   gridspec_kw=dict(width_ratios=[1, 1.25], wspace=0.30))

    bx = axm.boxplot([v_liv[m] for _, m, _ in GRP], positions=[0, 1], widths=0.6,
                     patch_artist=True, showfliers=False, medianprops=dict(color='black', lw=2.4),
                     whiskerprops=dict(lw=1.4), capprops=dict(lw=1.4))
    for b, (_, _, c) in zip(bx['boxes'], GRP):
        b.set(facecolor=c, alpha=0.75, edgecolor='black', lw=1.4)
    for i, (lab, m, c) in enumerate(GRP):
        vv = v_liv[m]
        axm.plot(i + rngs.normal(0, 0.055, len(vv)), vv, 'o', ms=6, color=c, mec='black',
                 mew=0.5, alpha=0.9)
    axm.axhline(0, color='0.35', lw=1.0, ls=':')
    top, bot = v_liv[is_pdac].max(), v_liv[is_pdac].min()
    span = top - bot
    ybar = top + 0.10 * span
    axm.plot([0, 0, 1, 1], [ybar, ybar + 0.025 * span, ybar + 0.025 * span, ybar], color='black', lw=1.4)
    axm.text(0.5, ybar + 0.045 * span, stars(u_met.pvalue), ha='center', va='bottom',
             fontsize=FS_ANNOT_LG)
    axm.set_ylim(bot - 0.08 * span, ybar + 0.21 * span)
    axm.set_xlim(-0.6, 1.6)
    axm.set_xticks([0, 1])
    axm.set_xticklabels([f'{l}\nn={int(m.sum())}' for l, m, _ in GRP], linespacing=1.4)
    axm.set_ylabel(YL)
    axm.text(0.035, 0.965, f'p = {u_met.pvalue:.2f}', transform=axm.transAxes,
             ha='left', va='top', fontsize=FS_ANNOT_LG)

    axc.plot(lca[m_ca], v_liv[m_ca], 'o', ms=7, color=C_CA, mec='black', mew=0.5, alpha=0.9)
    sl, ic, _, _ = theilslopes(v_liv[m_ca], lca[m_ca])
    xs = np.linspace(lca[m_ca].min(), lca[m_ca].max(), 50)
    axc.plot(xs, sl * xs + ic, color='0.35', lw=2.0, ls='--', zorder=1)
    axc.axhline(0, color='0.35', lw=1.0, ls=':')
    axc.text(0.035, 0.965, f'Spearman rho = {r_ca.statistic:+.2f}\np = {r_ca.pvalue:.2f}  '
             f'({stars(r_ca.pvalue)})\nn = {int(m_ca.sum())} patients', transform=axc.transAxes,
             ha='left', va='top', fontsize=FS_ANNOT_LG, linespacing=1.6)
    axc.text(0.965, 0.035, 'dashed: Theil-Sen fit', transform=axc.transAxes, ha='right', va='bottom',
             fontsize=FS_ANNOT_SM, color='0.45')
    axc.set_xlabel('log10(CA19-9 + 1)')
    axc.set_ylabel(YL)

    for a in (axm, axc):
        a.grid(axis='y', color='0.93', lw=0.8)
        a.set_axisbelow(True)
    axc.grid(axis='x', color='0.93', lw=0.8)
    fig.savefig(FIG / 'ladder_liver_independence.png', dpi=300, bbox_inches='tight')
    plt.show()
print(f'liver met {int(g_met.sum())} vs {int(g_no.sum())}: MWU p = {u_met.pvalue:.3f}')
print(f'CA19-9 within PDAC: rho {r_ca.statistic:+.3f}, p = {r_ca.pvalue:.3f}, n = {int(m_ca.sum())}')
print(f"-> {FIG / 'ladder_liver_independence.png'}  |  {LAD_DIR / 'liver_independence.csv'}")
```
